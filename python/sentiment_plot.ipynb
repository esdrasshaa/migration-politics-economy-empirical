```python
import pandas as pd
import matplotlib.pyplot as plt
import pandas as pd
```

## gpt-4.1-nano

```python
pd_gtp_sentiment = pd.read_csv("./gpt-4.1-nano/DATA/GptSentimentMonthlyVargptNano.csv")
```

```python
# 1. Handle index safely (convert to datetime and slice)
df_plot = pd_gtp_sentiment.copy()
if "month" in df_plot.columns:
    df_plot["month"] = pd.to_datetime(df_plot["month"])
    df_plot.set_index("month", inplace=True)
else:
    df_plot.index = pd.to_datetime(df_plot.index)

df_plot = df_plot.loc["1990-01-01":].sort_index()
df_plot_gpt = df_plot

# 2. Create the plot
fig, ax = plt.subplots(figsize=(14, 6))

# Plot 12-month rolling averages
df_plot["gpt_sentiment_government"].rolling(
    window=12, min_periods=3
).mean().plot(ax=ax, color="firebrick", linewidth=2.5, label="Government")

df_plot["gpt_sentiment_opposition"].rolling(
    window=12, min_periods=3
).mean().plot(ax=ax, color="steelblue", linewidth=2.5, label="Opposition")

df_plot["gtp_overall_sentiment"].rolling(
    window=12, min_periods=3
).mean().plot(
    ax=ax, color="gray", linewidth=1.8, linestyle=":", label="Overall"
)

# 3. Styling, formatting, and larger fonts
ax.axhline(0, color="black", linewidth=0.8, linestyle="--")

# Main Title & Subtitle mentioning the model
plt.suptitle(
    "Immigration Solidarity Net Score: Government vs. Opposition",
    fontsize=16,
    fontweight="bold",
    y=0.98,
)
ax.set_title(
    "12-Month Rolling Average | Sentiment computed with Gpt-4.1-nano",
    fontsize=12,
    color="dimgray",
    pad=10,
)

ax.set_xlabel("Year", fontsize=14)
ax.set_ylabel("Net Sentiment Score", fontsize=14)

# Tick labels
ax.tick_params(axis="both", which="major", labelsize=12)

# Legend font size
ax.legend(fontsize=12, loc="upper left")

# Model Badge / Watermark in the plot corner
ax.text(
    0.98,
    0.04,
    "Model: GPT-4.1-nano",
    transform=ax.transAxes,
    fontsize=10,
    color="gray",
    ha="right",
    va="bottom",
    bbox=dict(boxstyle="round,pad=0.3", facecolor="white", edgecolor="lightgray", alpha=0.8),
)

plt.tight_layout()
#plt.savefig("gov_vs_opposition_monthly.png", dpi=150)
plt.show()
```

### Doc2vec

```python
pd_sentiment = pd.read_csv("./results/immigration_sentiment_monthly_doc2vec.csv")
```

```python
# 1. Handle index safely (convert 'Date' to datetime and slice from 1990)
df_plot = pd_sentiment.copy()

if "Date" in df_plot.columns:
    df_plot["Date"] = pd.to_datetime(df_plot["Date"])
    df_plot.set_index("Date", inplace=True)
else:
    df_plot.index = pd.to_datetime(df_plot.index)

# Filter dates starting from 1990 and sort
df_plot = df_plot.loc["1990-01-01":].sort_index()
df_plot_doc2vec = df_plot

# 2. Create the plot
fig, ax = plt.subplots(figsize=(14, 6))

# Plot 12-month rolling averages
df_plot["sentiment_government"].rolling(
    window=12, min_periods=3
).mean().plot(ax=ax, color="firebrick", linewidth=2.5, label="Government")

df_plot["sentiment_opposition"].rolling(
    window=12, min_periods=3
).mean().plot(ax=ax, color="steelblue", linewidth=2.5, label="Opposition")

df_plot["sentiment_gesamt"].rolling(window=12, min_periods=3).mean().plot(
    ax=ax, color="gray", linewidth=1.8, linestyle=":", label="Overall"
)

# 3. Styling, formatting, and larger fonts
ax.axhline(0, color="black", linewidth=0.8, linestyle="--")

# Main Title & Subtitle mentioning Doc2Vec
plt.suptitle(
    "Immigration Solidarity Net Score: Government vs. Opposition",
    fontsize=16,
    fontweight="bold",
    y=0.98,
)
ax.set_title(
    "12-Month Rolling Average | Sentiment computed with Doc2Vec",
    fontsize=12,
    color="dimgray",
    pad=10,
)

ax.set_xlabel("Year", fontsize=14)
ax.set_ylabel("Net Sentiment Score", fontsize=14)

# Tick labels
ax.tick_params(axis="both", which="major", labelsize=12)

# Legend font size
ax.legend(fontsize=12, loc="upper left")

# Model Badge in the plot corner
ax.text(
    0.98,
    0.04,
    "Model: Doc2Vec",
    transform=ax.transAxes,
    fontsize=10,
    color="gray",
    ha="right",
    va="bottom",
    bbox=dict(
        boxstyle="round,pad=0.3",
        facecolor="white",
        edgecolor="lightgray",
        alpha=0.8,
    ),
)

plt.tight_layout()
plt.show()
```

### Doc2vec Vs GPT-4.1-nana

```python
# 2. Create the plot with a primary axis
fig, ax = plt.subplots(figsize=(14, 6))

# Plot GPT-4.1-nano on the primary (left) y-axis
df_plot_gpt["gtp_overall_sentiment"].rolling(
    window=12, min_periods=3
).mean().plot(
    ax=ax, color="firebrick", linewidth=1.8, label="GPT-4.1-nano"
)

# Plot GPT-5-nano on the primary (left) y-axis
df_plot_gpt5["gtp_overall_sentiment"].rolling(
    window=12, min_periods=3
).mean().plot(
    ax=ax, color="green", linewidth=1.8, linestyle =":", label="GPT-5-nano"
)

# Create a secondary y-axis sharing the same x-axis
ax2 = ax.twinx()

# Plot Doc2vec on the secondary (right) y-axis
df_plot_doc2vec["sentiment_gesamt"].rolling(
    window=12, min_periods=3
).mean().plot(
    ax=ax2, color="steelblue", linewidth=1.8,label="Doc2vec"
)

# 3. Styling, formatting, and larger fonts
ax.axhline(0, color="black", linewidth=0.8, linestyle="--")

# Main Title & Subtitle mentioning the model
plt.suptitle(
    "Immigration Solidarity Net Score Overall: Doc2vec vs.LLMs",
    fontsize=18,
    fontweight="bold",
    y=0.98,
)
ax.set_title(
    "12-Month Rolling Average",
    fontsize=14,
    color="dimgray",
    pad=10,
)

ax.set_xlabel("Year", fontsize=16)
ax.set_ylabel("GPT Net Sentiment Score", fontsize=18, color="firebrick")
ax2.set_ylabel("Doc2vec Net Sentiment Score", fontsize=18, color="steelblue")

# Tick labels for both axes
ax.tick_params(axis="both", which="major", labelsize=14)
ax2.tick_params(axis="both", which="major", labelsize=14)

# Combine legends from both axes into a single legend box
lines_1, labels_1 = ax.get_legend_handles_labels()
lines_2, labels_2 = ax2.get_legend_handles_labels()
ax.legend(lines_1 + lines_2, labels_1 + labels_2, fontsize=14, loc="upper left")

# Model Badge / Watermark in the plot corner
ax.text(
    0.98,
    0.04,
    "Model: Doc2vec vs LLMs",
    transform=ax.transAxes,
    fontsize=14,
    color="gray",
    ha="right",
    va="bottom",
    bbox=dict(boxstyle="round,pad=0.3", facecolor="white", edgecolor="lightgray", alpha=0.8),
)

plt.tight_layout()
#plt.savefig("gov_vs_opposition_monthly.png", dpi=150)
plt.show()
```
